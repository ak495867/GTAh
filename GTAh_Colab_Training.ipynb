# GTAH — Gabriel Transform Attention Hierarchy (Colab GPU Training)

This notebook trains a Mini-LM Transformer model comparing **Gabriel Transform Attention (GTAH)** against a standard **Dense Attention** baseline on Google Colab (with GPU acceleration).

### Features:
- 🚀 **GPU Acceleration** with Automatic Mixed Precision (`fp16` AMP)
- 📊 **Full Evaluation Metrics**: Cross-Entropy Loss, Perplexity, Token Accuracy, Peak VRAM, Inference Latency, and Throughput (tokens/sec)
- 📈 **Rich Visualizations**: Generates multi-panel plots saved directly as PNGs
- ⚖️ **Fair Controlled Benchmark**: Identical model architecture, parameter count, optimizer, learning rate schedule, and random seed
